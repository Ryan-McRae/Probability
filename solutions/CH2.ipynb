### a.) Find $P[H_1 \mid H_2]$
Recall the definition of conditional probability:
$$P[A \mid B] = \frac{P[AB]}{P[B]}$$

Applying this to $H_1$ and $H_2$:
$$P[H_1 \mid H_2] = \frac{P[H_1 H_2]}{P[H_2]} = \frac{\frac{1}{4} \times \frac{1}{4}}{\frac{1}{4}} = \frac{1}{4}$$

---

### b.) Find $P[H_1 \cap T_2]$
Following the branch for $H_1$ followed by $T_2$:
$$P[H_1 T_2] = P[H_1]\,P[T_2 \mid H_1] = \frac{1}{4} \times \frac{3}{4} = \frac{3}{16}$$

# Question 2.1.6

**Given:**
* Any person in the population has a $\frac{1}{5000}$ probability of being HIV-positive:
  $$P[H] = \frac{1}{5000} = 0.0002, \qquad P[N] = 1 - P[H] = \frac{4999}{5000} = 0.9998$$
* The testing kit gives the **correct** result $99\%$ of the time—meaning if you have HIV ($H$), you test positive ($+$), and if you do not have HIV ($N$), you test negative ($-$):
  $$P[+ \mid H] = 0.99 \qquad \text{and} \qquad P[- \mid N] = 0.99$$

---

### a.) False Negative Probability: $P[- \mid H]$
Since a person with HIV must either test positive or test negative, the false negative probability is the complement of $P[+ \mid H]$:
$$P[- \mid H] = 1 - P[+ \mid H] = 1 - 0.99 = 0.01 \quad (1\%)$$

---

### b.) Positive Predictive Value: $P[H \mid +]$
We want to find the probability of having HIV given a positive test result, $P[H \mid +]$. Using **Bayes' Theorem**:
$$P[H \mid +] = \frac{P[+ \mid H]\,P[H]}{P[+]}$$

To find the total probability of testing positive, $P[+]$, we apply the **Law of Total Probability** across the partition $\{H, N\}$ (using the false positive rate $P[+ \mid N] = 1 - P[- \mid N] = 0.01$):
$$P[+] = P[+ \cap H] + P[+ \cap N] = P[+ \mid H]\,P[H] + P[+ \mid N]\,P[N]$$

Substituting $P[+]$ back into Bayes' Theorem gives the full expression:
$$P[H \mid +] = \frac{P[+ \mid H]\,P[H]}{P[+ \mid H]\,P[H] + P[+ \mid N]\,P[N]}$$

Plugging in the given values:
$$P[H \mid +] = \frac{(0.99)\left(\frac{1}{5000}\right)}{(0.99)\left(\frac{1}{5000}\right) + (0.01)\left(\frac{4999}{5000}\right)} = \frac{0.99}{0.99 + 49.99} = \frac{99}{5098} \approx 0.0194 \quad (1.94\%)$$

To find $P[H_1 \mid D]$:
$$P[H_1 \mid D] = \frac{P[H_1 \cap D]}{P[D]}$$

Since every path in $D$ begins with $T_1$, the events $H_1$ and $D$ are **mutually exclusive** ($H_1 \cap D = \emptyset$), so:
$$P[H_1 \mid D] = \frac{0}{P[D]} = 0$$
